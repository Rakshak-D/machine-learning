## The following code is to analyze the life satisfaction rate of a country using GDP per capita (USD)

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
```

```python
data_root="../data/"
data=pd.read_csv(data_root+"lifesat/lifesat.csv")
```

```python
data
```

```python
gdp=data[["GDP per capita (USD)"]].values
satisfaction=data[['Life satisfaction']].values
```

```python
data.plot(kind="scatter",grid=True,x='GDP per capita (USD)',y='Life satisfaction')
plt.axis([23_500,62_500,4,9])
plt.show()
```

# Linear regression

```python
linear_regression_model=LinearRegression()
```

```python
linear_regression_model.fit(gdp,satisfaction)
```

```python
x_new=[[33_000]]
```

```python
print(linear_regression_model.predict(x_new))
```

# KNN

```python
knn_model=KNeighborsRegressor(n_neighbors=3)
```

```python
knn_model.fit(gdp,satisfaction)
```

```python
knn_model.predict(x_new)
```
